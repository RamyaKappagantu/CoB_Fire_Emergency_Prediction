## Severity Indicator Calculation- Property damage cost per fire incident.

```python
import pandas as pd

df = pd.read_excel('../Cleaned_dataset/new_DA.xlsx')
df.head()
```

```python
# Group by 'DAUID_new' and sum the 'Property_Damage_Cost'
total_damage_cost = df.groupby('DAUID_new')['Property_Loss_Value'].sum().reset_index()

# Group by 'DAUID_new' and count the number of incidents (each row is one incident)
total_fire_incidents = df.groupby('DAUID_new').size().reset_index(name='Fire_Incidents')

# Merge the two DataFrames on 'DAUID_new'
merged_df = pd.merge(total_damage_cost, total_fire_incidents, on='DAUID_new')

# Calculate the property damage cost per fire incident
merged_df['Damage_Cost_Per_Incident'] = (merged_df['Property_Loss_Value'] / merged_df['Fire_Incidents']).round(2)

# Rename the columns for clarity
merged_df.rename(columns={'DAUID_new': 'DAUID'}, inplace=True)

# Display the result
print(merged_df)

# Save the result to a CSV file
merged_df.to_csv('../Cleaned_dataset/property_damage_cost_per_fire_incident.csv', index=False, columns=['DAUID', 'Property_Loss_Value', 'Fire_Incidents', 'Damage_Cost_Per_Incident'])

print("Data saved to property_damage_cost_per_fire_incident.csv")
```
