## Severity Indicator Calculation- fire incident rate per 100 dwellings in each Dissemination Area 

To determine the residential fire incident rate per 100 dwellings in each dissemination area (DA) in Barrie, we would need two sets of data:

Number of residential fires in each DA: This data is typically collected by the local fire department or municipal authorities.
Number of residential dwellings in each DA: This data can be obtained from the Census data provided by Statistics Canada.

Here is a step-by-step approach to calculate the residential fire incident rate per 100 dwellings for each DA in Barrie:

* Step 1: Obtain the Data
Fire Incident Data: from CAD&RMS data on residential fire incidents by dissemination area.
Dwelling Count Data: Use the most recent Census data from Statistics Canada to get the number of residential dwellings in each DA in Barrie.

* Step 2: Compile the Data
Create a table where each row represents a dissemination area and includes the following columns:

DA Code
Number of Residential Fires
Number of Residential Dwellings

* Step 3: Calculate the Incident Rate
Use the following formula to calculate the residential fire incident rate per 100 dwellings for each DA:

## Finding Incident Percentage By DA

```python
# Group by 'DAUID' and count incidents
incident_counts = df.groupby('DAUID_new').size().reset_index(name='incident_count')

# Calculate the total number of incidents
total_incidents = incident_counts['incident_count'].sum()

# Calculate the incident percentage for each DAUID
incident_counts['incident_percentage'] = (incident_counts['incident_count'] / total_incidents) * 100

# Round the incident percentage to 2 decimal places
incident_counts['incident_percentage'] = incident_counts['incident_percentage'].round(2)
# Rename the column from 'DAUID_new' to 'DAUID'
incident_counts.rename(columns={'DAUID_new': 'DAUID'}, inplace=True)

# Save the DataFrame to a CSV file with proper column names
incident_counts.to_csv('../Cleaned_Dataset/incident_percentage_by_dguid.csv', index=False, columns=['DAUID', 'incident_count', 'incident_percentage'])

print("Data saved to incident_percentage_by_dguid.csv")
```
