```python
import pandas as pd

# Load the CSV file containing the mappings
mapping_csv_path = "../rawDataset/incident_types.csv"
df_mapping = pd.read_csv(mapping_csv_path)

# Convert the mapping DataFrame to a dictionary
incident_mapping = pd.Series(df_mapping.Description.values, index=df_mapping.Code).to_dict()

# Load the existing Excel file
excel_path = "../Cleaned_dataset/Aggregated_data.xlsx"
df_existing = pd.read_excel(excel_path)

# Map the incident categories to descriptions
df_existing["incident_category_description"] = df_existing["incident_category"].map(incident_mapping)

# Save the updated DataFrame to a new Excel file
updated_excel_path = "../Cleaned_dataset/updated_Aggregated_data.xlsx"
df_existing.to_excel(updated_excel_path, index=False)
updated_excel_path
```
