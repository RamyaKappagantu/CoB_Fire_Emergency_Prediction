```python
import pandas as pd
from IPython.display import display
import warnings
warnings.filterwarnings("ignore")

file_path = '../Cleaned_Dataset/updated_Aggregated_data.xlsx'
data = pd.read_excel(file_path)

display(data.head())
display(data.describe())
display(data.info())
```

```python
display(data.isnull().sum())
```

```python
#Basic Data Cleaning
import numpy as np

data['Received_Datetime'] = pd.to_datetime(data['Received_Datetime'], errors='coerce', format='%H:%M.%S')

# Drop rows where datetime conversion failed
data.dropna(subset=['Received_Datetime'], inplace=True)

# Convert incident type to numeric
data['inci_type'] = pd.to_numeric(data['inci_type'], errors='coerce')

# Drop rows with missing or zero geographic coordinates
data = data[(data['Longitude'] != 0) & (data['Latitude'] != 0)]
data.dropna(subset=['Longitude', 'Latitude'], inplace=True)

# Ensure no NaN values
data.replace([np.inf, -np.inf], np.nan, inplace=True)
data.dropna(inplace=True)

# Check for missing values again
print("Missing values after cleaning:\n", data.isnull().sum())
cleaned_data = data
```

```python
#Using prophet model

from sklearn.preprocessing import LabelEncoder
from prophet import Prophet
from sklearn.metrics import mean_squared_error, mean_absolute_error, r2_score
from sklearn.model_selection import ParameterGrid

df = cleaned_data

# Cleaning time data. There was .0 type format for seconds.

def clean_time_data(time_str):
    if isinstance(time_str, str):
        time_str = time_str.replace(".0", "")
        parts = time_str.split(':')
        hour = int(parts[0])
        if hour >= 24:
            hour = hour % 24
        return f"{hour:02d}:{parts[1]}"
    return time_str

df['Received_Datetime'] = df['Received_Datetime'].apply(clean_time_data)
df['Dispatched_Datetime'] = df['Dispatched_Datetime'].apply(clean_time_data)
df['Arrival_Datetime'] = df['Arrival_Datetime'].apply(clean_time_data)
df['Cleared_Datetime'] = df['Cleared_Datetime'].apply(clean_time_data)

# Formating date time column
df['Received_Datetime'] = pd.to_datetime(df['Received_Datetime'], format='%H:%M')
df['Dispatched_Datetime'] = pd.to_datetime(df['Dispatched_Datetime'], format='%H:%M')
df['Arrival_Datetime'] = pd.to_datetime(df['Arrival_Datetime'], format='%H:%M')
df['Cleared_Datetime'] = pd.to_datetime(df['Cleared_Datetime'], format='%H:%M')

# Encoding categorical features
label_encoders = {}
for column in ['incident_category_description', 'DAUID']:
    le = LabelEncoder()
    df[column] = le.fit_transform(df[column])
    label_encoders[column] = le

# Preparing the data for Prophet
df_prophet_dauid = df[['Received_Datetime', 'DAUID']].rename(columns={'Received_Datetime': 'ds', 'DAUID': 'y'})
df_prophet_category = df[['Received_Datetime', 'incident_category_description']].rename(columns={'Received_Datetime': 'ds', 'incident_category_description': 'y'})

# Splitting the data into training and testing sets (80 and 20)
train_size = int(0.8 * len(df))
train_dauid = df_prophet_dauid.iloc[:train_size]
test_dauid = df_prophet_dauid.iloc[train_size:]
train_category = df_prophet_category.iloc[:train_size]
test_category = df_prophet_category.iloc[train_size:]

test_dauid.to_csv('proph_test_dauid.csv', index=False)
test_category.to_csv('proph_test_category.csv', index=False)

# Hyperparameter tuning for DAUID
param_grid = {
    'changepoint_prior_scale': [0.01, 0.1, 0.5],
    'seasonality_prior_scale': [0.01, 0.1, 1.0],
    'holidays_prior_scale': [0.01, 0.1, 1.0]
}

best_mse_dauid = float('inf')
best_params_dauid = None

for params in ParameterGrid(param_grid):
    model_dauid = Prophet(**params)
    model_dauid.fit(train_dauid)
    forecast_dauid = model_dauid.predict(test_dauid[['ds']])
    mse = mean_squared_error(test_dauid['y'], forecast_dauid['yhat'])
    if mse < best_mse_dauid:
        best_mse_dauid = mse
        best_params_dauid = params

print(f'Best params for DAUID: {best_params_dauid}, Best MSE: {best_mse_dauid}')

# Training the best DAUID model
model_dauid = Prophet(**best_params_dauid)
model_dauid.fit(train_dauid)

# Hyperparameter tuning for Category
best_mse_category = float('inf')
best_params_category = None

for params in ParameterGrid(param_grid):
    model_category = Prophet(**params)
    model_category.fit(train_category)
    forecast_category = model_category.predict(test_category[['ds']])
    mse = mean_squared_error(test_category['y'], forecast_category['yhat'])
    if mse < best_mse_category:
        best_mse_category = mse
        best_params_category = params

print(f'Best params for Category: {best_params_category}, Best MSE: {best_mse_category}')

# Training the best Category model
model_category = Prophet(**best_params_category)
model_category.fit(train_category)

# Predicting DAUID
forecast_dauid = model_dauid.predict(test_dauid[['ds']])
predicted_dauid = forecast_dauid['yhat']
true_dauid = test_dauid['y'].values

# Predicting Category
forecast_category = model_category.predict(test_category[['ds']])
predicted_category = forecast_category['yhat']
true_category = test_category['y'].values

# Evaluation of the predictions
mse_dauid = mean_squared_error(true_dauid, predicted_dauid)
mae_dauid = mean_absolute_error(true_dauid, predicted_dauid)
r2_dauid = r2_score(true_dauid, predicted_dauid)

mse_category = mean_squared_error(true_category, predicted_category)
mae_category = mean_absolute_error(true_category, predicted_category)
r2_category = r2_score(true_category, predicted_category)

# Calculating directional accuracy for DAUID
test_dauid['yhat'] = predicted_dauid.values
test_dauid['correct_direction'] = np.sign(test_dauid['y'].diff().fillna(0)) == np.sign(test_dauid['yhat'].diff().fillna(0))
directional_accuracy_dauid = test_dauid['correct_direction'].mean()

# Calculating directional accuracy for Category
test_category['yhat'] = predicted_category.values
test_category['correct_direction'] = np.sign(test_category['y'].diff().fillna(0)) == np.sign(test_category['yhat'].diff().fillna(0))
directional_accuracy_category = test_category['correct_direction'].mean()

print(f'DAUID - MSE: {mse_dauid}, MAE: {mae_dauid}, R2: {r2_dauid}, Directional Accuracy: {directional_accuracy_dauid}')
print(f'Category - MSE: {mse_category}, MAE: {mae_category}, R2: {r2_category}, Directional Accuracy: {directional_accuracy_category}')

# Function to predict the next DAUID and Category
def predict_next_dauid_and_category(model_dauid, model_category, label_encoders):
    # Create a future dataframe for the next prediction
    future_dauid = model_dauid.make_future_dataframe(periods=1, freq='H', include_history=False)
    future_category = model_category.make_future_dataframe(periods=1, freq='H', include_history=False)
    
    # Predicting the next DAUID and Category
    pred_dauid = model_dauid.predict(future_dauid)['yhat'].iloc[-1]
    pred_category = model_category.predict(future_category)['yhat'].iloc[-1]
    
    # Get the predicted class labels
    pred_dauid_label = label_encoders['DAUID'].inverse_transform([int(pred_dauid)])[0]
    pred_category_label = label_encoders['incident_category_description'].inverse_transform([int(pred_category)])[0]
    
    return pred_dauid_label, pred_category_label

# Predict the next DAUID and Category
next_dauid, next_category = predict_next_dauid_and_category(model_dauid, model_category, label_encoders)

print(f'Next DAUID: {next_dauid}')
print(f'Next Category: {next_category}')
```
